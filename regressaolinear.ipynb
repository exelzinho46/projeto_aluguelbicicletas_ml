```python
from sklearn.linear_model import LinearRegression

# Orçamento de marketing (recursos - X)
x = [[1000], [2500], [3250], [5000], [6500]]

# Lucro em vendas (alvo - y)
y = [10000, 20500, 33250, 59000, 77000]

modelo = LinearRegression()

# Correção feita aqui: mudado de (y, x) para (x, y)
modelo.fit(x, y)

predicao = modelo.predict([[20000]])

print(f"Predição de lucro: R$ {predicao[]:.2f}")
```
